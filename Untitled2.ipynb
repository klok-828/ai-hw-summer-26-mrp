```python

# %% [Cell 1] Setup
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
import time
import matplotlib.pyplot as plt

print("TF version:", tf.__version__)
print("GPU available:", tf.config.list_physical_devices('GPU'))

# %% [Cell 2] Load data (CIFAR-10)
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()
y_train = y_train.flatten()
y_test = y_test.flatten()

# Use a subset for a fast demo run. Bump these up for a more rigorous result.
N_TRAIN = 10000
N_TEST = 2000
x_train, y_train = x_train[:N_TRAIN], y_train[:N_TRAIN]
x_test, y_test = x_test[:N_TEST], y_test[:N_TEST]

x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

print(f"Train: {x_train.shape}, Test: {x_test.shape}")

# %% [Cell 3] Resize helper
def resize_dataset(images, size):
    """Resize a batch of images to (size, size). Native CIFAR-10 is 32x32,
    so 64 and 128 are UPSCALED — this isolates 'resolution' as the variable
    without changing the underlying image content or dataset."""
    resized = tf.image.resize(images, (size, size), method='bilinear')
    return resized.numpy()

# %% [Cell 4] Model builder — SAME architecture at every resolution
def build_cnn(input_shape, num_classes=10):
    model = models.Sequential([
        layers.Input(shape=input_shape),
        layers.Conv2D(32, 3, padding='same', activation='relu'),
        layers.MaxPooling2D(2),
        layers.Conv2D(64, 3, padding='same', activation='relu'),
        layers.MaxPooling2D(2),
        layers.Conv2D(128, 3, padding='same', activation='relu'),
        layers.MaxPooling2D(2),
        # GlobalAveragePooling instead of Flatten keeps param count constant
        # across resolutions, so the comparison isolates resolution as the
        # true variable rather than "how big did the flattened vector get".
        layers.GlobalAveragePooling2D(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(num_classes, activation='softmax'),
    ])
    model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model

# %% [Cell 5] Run the experiment across resolutions
resolutions = [32, 64, 128]
EPOCHS = 8
BATCH_SIZE = 64

results = {}

for res in resolutions:
    print(f"\n{'='*50}\nResolution: {res}x{res}\n{'='*50}")

    x_tr = resize_dataset(x_train, res)
    x_te = resize_dataset(x_test, res)

    model = build_cnn((res, res, 3))
    n_params = model.count_params()

    start = time.time()
    history = model.fit(x_tr, y_train,
                         validation_data=(x_te, y_test),
                         epochs=EPOCHS,
                         batch_size=BATCH_SIZE,
                         verbose=2)
    train_time = time.time() - start

    # Inference timing over the full test set
    start = time.time()
    _ = model.predict(x_te, batch_size=BATCH_SIZE, verbose=0)
    infer_time = time.time() - start

    test_loss, test_acc = model.evaluate(x_te, y_test, verbose=0)

    results[res] = {
        'test_accuracy': test_acc,
        'train_time_sec': train_time,
        'inference_time_sec': infer_time,
        'inference_time_per_image_ms': (infer_time / len(x_te)) * 1000,
        'params': n_params,
        'history': history.history,
    }

    print(f"Test accuracy: {test_acc:.4f}")
    print(f"Params: {n_params:,}")
    print(f"Training time: {train_time:.1f}s")
    print(f"Inference time (full test set): {infer_time:.2f}s "
          f"({results[res]['inference_time_per_image_ms']:.3f} ms/image)")

# %% [Cell 6] Summary table
print(f"\n{'Resolution':<12}{'Accuracy':<12}{'Train Time (s)':<18}{'Infer (ms/img)':<16}{'Params':<12}")
for res in resolutions:
    r = results[res]
    print(f"{str(res)+'x'+str(res):<12}{r['test_accuracy']:<12.4f}"
          f"{r['train_time_sec']:<18.1f}{r['inference_time_per_image_ms']:<16.3f}"
          f"{r['params']:<12,}")

# %% [Cell 7] Plots
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

accs = [results[r]['test_accuracy'] for r in resolutions]
times = [results[r]['train_time_sec'] for r in resolutions]
infers = [results[r]['inference_time_per_image_ms'] for r in resolutions]

axes[0].plot(resolutions, accs, marker='o')
axes[0].set_title('Test Accuracy vs Resolution')
axes[0].set_xlabel('Resolution (px)')
axes[0].set_ylabel('Accuracy')
axes[0].grid(alpha=0.3)

axes[1].plot(resolutions, times, marker='o', color='orange')
axes[1].set_title('Training Time vs Resolution')
axes[1].set_xlabel('Resolution (px)')
axes[1].set_ylabel('Time (s)')
axes[1].grid(alpha=0.3)

axes[2].plot(resolutions, infers, marker='o', color='green')
axes[2].set_title('Inference Time per Image vs Resolution')
axes[2].set_xlabel('Resolution (px)')
axes[2].set_ylabel('Time (ms/image)')
axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.savefig('resolution_comparison.png', dpi=150)
plt.show()

# %% [Cell 8] (Optional) Training curves per resolution
fig, ax = plt.subplots(figsize=(7, 5))
for res in resolutions:
    ax.plot(results[res]['history']['val_accuracy'], label=f'{res}x{res}')
ax.set_title('Validation Accuracy over Epochs, by Resolution')
ax.set_xlabel('Epoch')
ax.set_ylabel('Validation Accuracy')
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig('accuracy_curves.png', dpi=150)
plt.show()
```
